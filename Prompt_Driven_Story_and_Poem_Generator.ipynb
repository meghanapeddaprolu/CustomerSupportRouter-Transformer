```python
from google import genai
from google.colab import userdata

# Read the Gemini API key from Colab Secrets
API_KEY = userdata.get("GEMINI_API_KEY")

# Create the Gemini client
client = genai.Client(api_key=API_KEY)

print("✅ Gemini API configured successfully!")
```

```python
response = client.models.generate_content(
    model="models/gemini-flash-latest",
    contents="Say hello in one sentence."
)

print(response.text)
```

```python
# Gemini model to use throughout the notebook
MODEL_NAME = "models/gemini-flash-latest"
```

```python
def generate_text(prompt, temperature=0.7, top_p=0.9):
    response = client.models.generate_content(
        model=MODEL_NAME,
        contents=prompt,
        config={
            "temperature": temperature,
            "top_p": top_p
        }
    )

    return response.text
```

```python
prompt = "Write a short motivational quote about success."

output = generate_text(prompt)

print(output)
```

```python
topic = "A lonely tree standing in the middle of a desert"

poet_prompt = f"""
You are a famous poet.

Write a beautiful and emotional poem about {topic}.

Requirements:
- 12 to 16 lines
- Use vivid imagery and symbolism.
- Make the reader feel hope and loneliness.
- Use expressive and artistic language.
"""

poet_output = generate_text(poet_prompt)

print("===== POET OUTPUT =====\n")
print(poet_output)
```

```python
journalist_prompt = f"""
You are an experienced journalist.

Write a news article about {topic}.

Requirements:
- Give the article a suitable headline.
- Use a formal and objective tone.
- Keep it around 200 words.
- Explain why the tree is important and include possible environmental significance.
"""

journalist_output = generate_text(journalist_prompt)

print("===== JOURNALIST OUTPUT =====\n")
print(journalist_output)
```

```python
teacher_prompt = f"""
You are an experienced school teacher.

Explain the importance of {topic} to students.

Requirements:
- Use simple and easy-to-understand language.
- Keep it around 150–200 words.
- Explain why trees are important in deserts.
- Include an environmental lesson at the end.
- Make the explanation educational and engaging.
"""

teacher_output = generate_text(teacher_prompt)

print("===== TEACHER OUTPUT =====\n")
print(teacher_output)
```
